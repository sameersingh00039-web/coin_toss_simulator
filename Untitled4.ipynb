```python
"""
Project 33: Coin Toss Simulator
Generates heads or tails randomly using toss_coin().
"""

import random


def toss_coin():
    """Randomly return 'Heads' or 'Tails'."""
    return random.choice(["Heads", "Tails"])


def main():
    print("=== Coin Toss Simulator ===")
    heads = 0
    tails = 0

    while True:
        choice = input("\nPress Enter to toss the coin (or type 'q' to quit): ").strip().lower()
        if choice == "q":
            break

        result = toss_coin()
        print(f"Result: {result}")

        if result == "Heads":
            heads += 1
        else:
            tails += 1

    total = heads + tails
    print("\n--- Summary ---")
    print(f"Total tosses: {total}")
    print(f"Heads: {heads}")
    print(f"Tails: {tails}")
    print("Thanks for playing!")


if __name__ == "__main__":
    main()
```
